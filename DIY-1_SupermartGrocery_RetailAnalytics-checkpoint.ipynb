## 1. Key Performance Indicators (KPIs)

## Key Performance Indicators (KPIs) for The Shoppe

Based on the project requirements and general retail analytics best practices, the following 10+ Key Performance Indicators (KPIs) have been identified as crucial for "The Shoppe" to analyze its sales, understand customer demands, and plan for national expansion:

1.  **Total Sales Revenue:** This fundamental KPI measures the overall income generated from sales over a specific period. It provides a direct indication of the business's top-line performance.

2.  **Average Transaction Value (ATV):** Calculated as Total Sales Revenue divided by the Number of Transactions, ATV indicates the average amount spent by a customer in a single purchase. Increasing ATV is often a key objective for retailers.

3.  **Customer Acquisition Rate:** This KPI measures the rate at which "The Shoppe" is attracting new customers. It can be calculated by dividing the number of new customers acquired in a period by the total number of customers at the end of that period (or by tracking unique customer IDs over time).

4.  **Customer Retention Rate:** This measures the percentage of existing customers that continue to shop at "The Shoppe" over a specific period. High retention rates are generally more cost-effective than acquiring new customers.

5.  **Sales Growth Rate:** This KPI tracks the percentage increase or decrease in sales revenue over a defined period (e.g., month-over-month, year-over-year). It is a critical indicator of business growth momentum.

6.  **Profit Margin per Product/Category/Sub-Category:** This measures the profitability of individual products, categories, or sub-categories after accounting for the cost of goods sold. It helps identify which items contribute most to the bottom line.

7.  **Inventory Turnover Rate:** Calculated as the Cost of Goods Sold divided by the Average Inventory, this KPI indicates how many times inventory is sold and replaced over a period. A higher turnover generally suggests efficient inventory management, though it varies by product type.

8.  **Top Selling Products/Categories/Sub-Categories:** Identifying the products, categories, or sub-categories that generate the highest sales volume or revenue. This helps in stocking, marketing, and promotional decisions.

9.  **Sales by Region/State:** This KPI breaks down sales performance by different geographical areas (e.g., Tamil Nadu initially, and then other states as the business expands). It is essential for understanding regional preferences and tailoring strategies accordingly.

10. **Customer Lifetime Value (CLTV):** This metric predicts the total net profit a business can expect to earn from a customer throughout their entire relationship. It helps in understanding the long-term value of customers and guiding marketing spend.

11. **Basket Size (Average Items per Transaction):** This is the average number of distinct items purchased in a single transaction. It provides insights into purchasing patterns and can be a target for up-selling or cross-selling initiatives.

These KPIs will form the basis for the subsequent data analysis, helping to derive actionable insights for "The Shoppe".

### 2.2. Outlier Detection

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
sns.boxplot(x=df["Sales"])
plt.title("Boxplot of Sales")
plt.show()
# Load the preprocessed data (assuming it was saved, or re-run preprocessing steps)
# For this script, we'll assume the preprocessed_supermart_sales.csv is available
# If not, we would load the original and re-apply preprocessing.
# df = pd.read_csv("C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/Supermart Grocery Sales - Retail Analytics Dataset.csv", parse_dates=["Order Date"])

# As the previous script did not save the df to a CSV, we will reload and preprocess again for this isolated script.
# In a notebook, the df would persist in memory.
df = pd.read_csv("C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/Supermart Grocery Sales - Retail Analytics Dataset.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].str.strip()

print("Dataset loaded and basic preprocessing applied for outlier detection.")

# --- Outlier Detection --- #

# We will focus on numerical columns: 'Sales' and 'Profit'
numerical_cols = ["Sales", "Profit"]

plt.style.use("seaborn-v0_8-whitegrid")  # Seaborn style for clean visuals

# Define plot and summary output directories
import os
output_plot_dir = "C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/"
if not os.path.exists(output_plot_dir):
    os.makedirs(output_plot_dir)

# Create proper text file for summary
outlier_summary_file = os.path.join(output_plot_dir, "outlier_summary.txt")

with open(outlier_summary_file, "w") as f_summary:
    f_summary.write("Outlier Detection Summary:\n\n")

    for col in numerical_cols:
        f_summary.write(f"--- Analyzing column: {col} ---\n")
        print(f"\nAnalyzing column: {col}")

        # 1. Visualization with Boxplot
        plt.figure(figsize=(8, 6))
        sns.boxplot(x=df[col])
        plt.title(f"Boxplot of {col}")
        boxplot_path = os.path.join(output_plot_dir, f"{col}_boxplot.png")
        plt.savefig(boxplot_path)
        plt.close()
        print(f"Boxplot for {col} saved to {boxplot_path}")
        f_summary.write(f"Boxplot saved to: {boxplot_path}\n")

        # 2. IQR (Interquartile Range) Method
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        f_summary.write(f"Q1 for {col}: {Q1}\n")
        f_summary.write(f"Q3 for {col}: {Q3}\n")
        f_summary.write(f"IQR for {col}: {IQR}\n")
        f_summary.write(f"Lower bound for {col}: {lower_bound}\n")
        f_summary.write(f"Upper bound for {col}: {upper_bound}\n")

        outliers_iqr = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        f_summary.write(f"Number of outliers detected in {col} using IQR: {len(outliers_iqr)}\n")
        if not outliers_iqr.empty:
            f_summary.write(f"Summary stats of outliers in {col}:\n{outliers_iqr[[col]].describe()}\n")
        else:
            f_summary.write(f"No outliers detected in {col} based on IQR method.\n")

        print(f"Number of outliers in {col} (IQR method): {len(outliers_iqr)}\n")

    f_summary.write("Outlier detection complete. No changes made to the dataset at this stage.\n")

print(f"\n Outlier detection completed.")
print(f"Summary saved at: {outlier_summary_file}")
print(f"Plots saved in: {output_plot_dir}")

```

## 3. Exploratory Data Analysis (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define paths
eda_plot_dir = "C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/"
eda_summary_path = os.path.join(eda_plot_dir, "eda_summary.txt")

# Create directory if it doesn't exist
os.makedirs(eda_plot_dir, exist_ok=True)

# Load dataset
df = pd.read_csv(os.path.join(eda_plot_dir, "Supermart Grocery Sales - Retail Analytics Dataset.csv"))
df["Order Date"] = pd.to_datetime(df["Order Date"])
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].str.strip()

print("Dataset loaded and basic preprocessing applied for EDA.")

# Begin EDA Summary
with open(eda_summary_path, "w") as f_summary:
    f_summary.write("Exploratory Data Analysis (EDA) Summary:\n\n")
    plt.style.use("seaborn-v0_8-whitegrid")

    # 1. Total Sales Revenue
    total_revenue = df["Sales"].sum()
    f_summary.write(f"1. Total Sales Revenue: {total_revenue:,.2f}\n")

    # 2. Average Transaction Value
    atv = df["Sales"].mean()
    f_summary.write(f"2. Average Transaction Value (ATV): {atv:,.2f}\n")

    # Sales Distribution Plot
    plt.figure(figsize=(10, 6))
    sns.histplot(df["Sales"], kde=True, bins=50)
    plt.title("Distribution of Sales per Transaction")
    plt.xlabel("Sales Amount")
    plt.ylabel("Frequency")
    sales_dist_path = os.path.join(eda_plot_dir, "sales_distribution.png")
    plt.savefig(sales_dist_path)
    plt.show()
    f_summary.write(f"   - Sales Distribution Plot: {sales_dist_path}\n")

    # 3. Customer Analysis
    num_customers = df["Customer Name"].nunique()
    f_summary.write(f"3. Number of Unique Customers: {num_customers}\n")
    customer_counts = df["Customer Name"].value_counts()
    f_summary.write(f"   - Top 5 Most Frequent Customers:\n{customer_counts.head().to_string()}\n")
    f_summary.write(f"   - Avg Transactions per Customer: {df.shape[0]/num_customers:.2f}\n")

    plt.figure(figsize=(12, 6))
    customer_counts.head(10).plot(kind="bar")
    plt.title("Top 10 Customers by Number of Transactions")
    plt.xlabel("Customer Name")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    top_customers_path = os.path.join(eda_plot_dir, "top_10_customers_by_transactions.png")
    plt.savefig(top_customers_path)
    plt.show()
    f_summary.write(f"   - Top 10 Customers Plot: {top_customers_path}\n")

    # 4. Monthly Sales Trend
    df["Order Month"] = df["Order Date"].dt.to_period("M")
    monthly_sales = df.groupby("Order Month")["Sales"].sum()
    monthly_sales.index = monthly_sales.index.to_timestamp()

    plt.figure(figsize=(12, 6))
    monthly_sales.plot(marker="o")
    plt.title("Monthly Sales Revenue Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales Revenue")
    plt.xticks(rotation=45)
    plt.tight_layout()
    monthly_sales_path = os.path.join(eda_plot_dir, "monthly_sales_trend.png")
    plt.savefig(monthly_sales_path)
    plt.show()
    f_summary.write(f"4. Monthly Sales Trend Plot: {monthly_sales_path}\n")
    f_summary.write(f"   - Monthly Sales Data:\n{monthly_sales.to_string()}\n")

    # 5. Profit Analysis
    total_profit = df["Profit"].sum()
    avg_profit = df["Profit"].mean()
    df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100
    avg_margin = df["Profit Margin"].mean()
    f_summary.write(f"5. Profit Analysis:\n")
    f_summary.write(f"   - Total Profit: {total_profit:,.2f}\n")
    f_summary.write(f"   - Average Profit/Transaction: {avg_profit:,.2f}\n")
    f_summary.write(f"   - Average Profit Margin: {avg_margin:.2f}%\n")

    # Profit Distribution
    plt.figure(figsize=(10, 6))
    sns.histplot(df["Profit"], kde=True, bins=50)
    plt.title("Profit Distribution")
    plt.xlabel("Profit")
    plt.ylabel("Frequency")
    profit_path = os.path.join(eda_plot_dir, "profit_distribution.png")
    plt.savefig(profit_path)
    plt.show()
    f_summary.write(f"   - Profit Distribution Plot: {profit_path}\n")

    # Profit Margin Distribution
    plt.figure(figsize=(10, 6))
    sns.histplot(df["Profit Margin"].dropna(), kde=True, bins=50)
    plt.title("Profit Margin Distribution")
    plt.xlabel("Profit Margin (%)")
    plt.ylabel("Frequency")
    margin_path = os.path.join(eda_plot_dir, "profit_margin_distribution.png")
    plt.savefig(margin_path)
    plt.show()
    f_summary.write(f"   - Profit Margin Distribution Plot: {margin_path}\n")

    # Profit Margin by Category
    margin_by_cat = df.groupby("Category")["Profit Margin"].mean().sort_values(ascending=False)
    plt.figure(figsize=(12, 7))
    margin_by_cat.plot(kind="bar")
    plt.title("Average Profit Margin by Category")
    plt.xlabel("Category")
    plt.ylabel("Profit Margin (%)")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    margin_cat_path = os.path.join(eda_plot_dir, "profit_margin_by_category.png")
    plt.savefig(margin_cat_path)
    plt.show()
    f_summary.write(f"   - Profit Margin by Category Plot: {margin_cat_path}\n")
    f_summary.write(f"   - Profit Margin Data:\n{margin_by_cat.to_string()}\n")

    # 6. Top Selling & Profitable Categories/Sub-Categories
    sales_cat = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
    sales_subcat = df.groupby("Sub Category")["Sales"].sum().sort_values(ascending=False)
    profit_cat = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
    profit_subcat = df.groupby("Sub Category")["Profit"].sum().sort_values(ascending=False)

    f_summary.write("6. Category/Sub-Category Sales & Profit:\n")
    f_summary.write(f"   - Sales by Category:\n{sales_cat.to_string()}\n")
    f_summary.write(f"   - Top 5 Sub-Categories (Sales):\n{sales_subcat.head().to_string()}\n")
    f_summary.write(f"   - Profit by Category:\n{profit_cat.to_string()}\n")
    f_summary.write(f"   - Top 5 Sub-Categories (Profit):\n{profit_subcat.head().to_string()}\n")

    fig, axs = plt.subplots(2, 2, figsize=(18, 12))
    sales_cat.plot(kind="bar", ax=axs[0, 0], title="Sales by Category")
    sales_subcat.head(10).plot(kind="bar", ax=axs[0, 1], title="Top 10 Selling Sub-Categories")
    profit_cat.plot(kind="bar", ax=axs[1, 0], title="Profit by Category")
    profit_subcat.head(10).plot(kind="bar", ax=axs[1, 1], title="Top 10 Profitable Sub-Categories")
    for ax in axs.flat:
        ax.tick_params(axis="x", rotation=45)
    plt.tight_layout()
    cat_subcat_path = os.path.join(eda_plot_dir, "category_subcategory_sales_profit.png")
    plt.savefig(cat_subcat_path)
    plt.show()
    f_summary.write(f"   - Combined Plot: {cat_subcat_path}\n")

    # 7. Geographical Sales
    sales_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
    sales_city = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

    f_summary.write("7. Sales by Geography:\n")
    f_summary.write(f"   - Sales by Region:\n{sales_region.to_string()}\n")
    f_summary.write(f"   - Top 5 Cities:\n{sales_city.head().to_string()}\n")

    fig, axs = plt.subplots(1, 2, figsize=(18, 6))
    sales_region.plot(kind="bar", ax=axs[0], title="Sales by Region")
    sales_city.head(10).plot(kind="bar", ax=axs[1], title="Top 10 Cities")
    for ax in axs:
        ax.tick_params(axis="x", rotation=45)
    plt.tight_layout()
    geo_path = os.path.join(eda_plot_dir, "geographical_sales.png")
    plt.savefig(geo_path)
    plt.show()
    f_summary.write(f"   - Geography Plot: {geo_path}\n")

    # 8. Discount Analysis
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x="Discount", y="Sales", alpha=0.5)
    plt.title("Sales vs Discount")
    discount_sales_path = os.path.join(eda_plot_dir, "discount_vs_sales.png")
    plt.savefig(discount_sales_path)
    plt.show()
    f_summary.write(f"8. Discount vs Sales Plot: {discount_sales_path}\n")

    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x="Discount", y="Profit", alpha=0.5)
    plt.title("Profit vs Discount")
    discount_profit_path = os.path.join(eda_plot_dir, "discount_vs_profit.png")
    plt.savefig(discount_profit_path)
    plt.show()
    f_summary.write(f"   - Discount vs Profit Plot: {discount_profit_path}\n")

    # 9. Correlation Matrix
    numeric_df = df[["Sales", "Discount", "Profit", "Profit Margin"]].replace([float("inf"), -float("inf")], pd.NA).dropna()
    corr_matrix = numeric_df.corr()

    plt.figure(figsize=(8, 6))
    sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Correlation Matrix")
    corr_path = os.path.join(eda_plot_dir, "correlation_matrix.png")
    plt.savefig(corr_path)
    plt.show()
    f_summary.write(f"9. Correlation Matrix Plot: {corr_path}\n")
    f_summary.write(f"   - Correlation Matrix:\n{corr_matrix.to_string()}\n")

    f_summary.write("\nEDA complete. All visuals and insights saved.\n")

print(f"\nExploratory Data Analysis completed.\nSummary saved to: {eda_summary_path}\nPlots saved in: {eda_plot_dir}")
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
import os

# Create directory for association rule results if it doesn't exist
results_dir = r"C:\Users\Shank\Downloads\New folder\DIY_1_RetailSupermarket"
if not os.path.exists(results_dir):
    os.makedirs(results_dir)

association_summary_path = os.path.join(results_dir, "association_rules_summary.txt")
association_rules_csv_path = os.path.join(results_dir, "association_rules.csv")
frequent_itemsets_csv_path = os.path.join(results_dir, "frequent_itemsets.csv")

# Load the dataset
df = pd.read_csv(r"C:\Users\Shank\Downloads\New folder\DIY_1_RetailSupermarket\Supermart Grocery Sales - Retail Analytics Dataset.csv")

print("Preparing data for association rule mining...")

# Create basket: rows=transactions, columns=items, values=counts of 'Sales' per item per order
df_basket = df.groupby(["Order ID", "Sub Category"])["Sales"].count().unstack().fillna(0)

# Encode basket as boolean: True if item bought in transaction, else False
df_basket_encoded = df_basket > 0

# Remove any transactions with no items (all False)
df_basket_encoded = df_basket_encoded[df_basket_encoded.any(axis=1)]

if df_basket_encoded.empty:
    print("Basket is empty after encoding. Cannot proceed with association rule mining.")
    with open(association_summary_path, "w") as f_summary:
        f_summary.write("Association Rule Mining Summary:\n\n")
        f_summary.write("Error: Basket is empty after encoding. Cannot perform association rule mining.\n")
else:
    print(f"Basket prepared with {df_basket_encoded.shape[0]} transactions and {df_basket_encoded.shape[1]} unique items.")
    
    min_support_value = 0.005  # Lowered minimum support to 0.5%
    print(f"Applying Apriori algorithm with min_support = {min_support_value}...")
    frequent_itemsets = apriori(df_basket_encoded, min_support=min_support_value, use_colnames=True)

    if frequent_itemsets.empty:
        print("No frequent itemsets found with the current min_support. Try lowering it further.")
        with open(association_summary_path, "w") as f_summary:
            f_summary.write("Association Rule Mining Summary:\n\n")
            f_summary.write(f"No frequent itemsets found with min_support = {min_support_value}.\n")
            f_summary.write("Consider lowering the min_support threshold.\n")
    else:
        print(f"Found {len(frequent_itemsets)} frequent itemsets.")
        frequent_itemsets.to_csv(frequent_itemsets_csv_path, index=False)
        print(f"Frequent itemsets saved to {frequent_itemsets_csv_path}")

        min_confidence_value = 0.05  # Lowered minimum confidence to 5%
        print(f"Generating association rules with min_confidence = {min_confidence_value}...")
        rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=min_confidence_value)

        if rules.empty:
            print("No association rules found with the current confidence threshold.")
            with open(association_summary_path, "w") as f_summary:
                f_summary.write("Association Rule Mining Summary:\n\n")
                f_summary.write(f"Frequent itemsets found: {len(frequent_itemsets)}\n")
                f_summary.write(frequent_itemsets.to_string() + "\n\n")
                f_summary.write(f"No association rules generated with min_confidence = {min_confidence_value}.\n")
                f_summary.write("Consider lowering the confidence threshold or adjusting min_support.\n")
        else:
            print(f"Found {len(rules)} association rules.")
            rules_sorted = rules.sort_values(["lift", "confidence"], ascending=[False, False])
            rules_sorted.to_csv(association_rules_csv_path, index=False)
            print(f"Association rules saved to {association_rules_csv_path}")

            with open(association_summary_path, "w") as f_summary:
                f_summary.write("Association Rule Mining Summary:\n\n")
                f_summary.write("Parameters used:\n")
                f_summary.write(f"  - Minimum Support: {min_support_value}\n")
                f_summary.write(f"  - Minimum Confidence: {min_confidence_value}\n\n")
                f_summary.write(f"Number of frequent itemsets found: {len(frequent_itemsets)}\n")
                f_summary.write("Top 10 Frequent Itemsets (by support):\n")
                f_summary.write(frequent_itemsets.sort_values('support', ascending=False).head(10).to_string() + "\n\n")
                f_summary.write(f"Number of association rules generated: {len(rules_sorted)}\n")
                f_summary.write("Top 10 Association Rules (by lift, then confidence):\n")
                f_summary.write(rules_sorted.head(10).to_string() + "\n")

print(f"Association rule mining process complete. Summary saved to {association_summary_path}")
```

**Summary of Association Rule Mining:**

    - The data was transformed into a one-hot encoded basket format.
    - Frequent itemsets were generated with a minimum support of 0.01.

## 5. Classification Modeling

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score
import os

# Create directory for classification results if it doesn't exist
results_dir = "C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/classification_results"
if not os.path.exists(results_dir):
    os.makedirs(results_dir)

classification_summary_path = os.path.join(results_dir, "classification_summary.txt")
predicted_high_performers_path = os.path.join(results_dir, "predicted_high_performers.csv")

# Load the dataset
df = pd.read_csv("C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/Supermart Grocery Sales - Retail Analytics Dataset.csv") 

# Basic preprocessing (strip whitespace, convert date)
df["Order Date"] = pd.to_datetime(df["Order Date"])
for col in df.select_dtypes(include=["object"]).columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna("Unknown")
    df[col] = df[col].str.strip()

print("Dataset loaded and basic preprocessing applied.")

# Aggregate data for classification
df_agg = df.groupby(["Region", "Category", "Sub Category"]).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

print(f"Aggregated data created with {len(df_agg)} unique (Region, Category, Sub Category) combinations.")

# Define target variables: top 25% per region for sales and profit
df_agg["Is_High_Sales_Generator"] = 0
df_agg["Is_High_Profit_Generator"] = 0

for region in df_agg["Region"].unique():
    region_sales = df_agg[df_agg["Region"] == region]["Total_Sales"]
    region_profit = df_agg[df_agg["Region"] == region]["Total_Profit"]

    if len(region_sales) > 0:
        sales_threshold = region_sales.quantile(0.75)
        df_agg.loc[(df_agg["Region"] == region) & (df_agg["Total_Sales"] >= sales_threshold), "Is_High_Sales_Generator"] = 1
    
    if len(region_profit) > 0:
        profit_threshold = region_profit.quantile(0.75)
        df_agg.loc[(df_agg["Region"] == region) & (df_agg["Total_Profit"] >= profit_threshold), "Is_High_Profit_Generator"] = 1

print("Target variables (Is_High_Sales_Generator, Is_High_Profit_Generator) created.")

# Prepare features and preprocessing pipeline
features = ["Region", "Category", "Sub Category"]
X = df_agg[features]

categorical_features = ["Region", "Category", "Sub Category"]
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features)
    ],
    remainder="passthrough"
)

# Train models for sales and profit classification
results_summary = []
all_predictions_list = []

for target_variable_name, target_series in [
    ("Is_High_Sales_Generator", df_agg["Is_High_Sales_Generator"]),
    ("Is_High_Profit_Generator", df_agg["Is_High_Profit_Generator"])
]:
    results_summary.append(f"--- Classification Model for: {target_variable_name} ---")
    print(f"\nTraining model for: {target_variable_name}")

    y = target_series

    if len(y.unique()) < 2:
        print(f"Skipping {target_variable_name}: Target variable has only one class.")
        results_summary.append(f"Skipped: Target variable has only one class ({y.unique()}).\n")
        continue
    if len(X) < 10:
        print(f"Skipping {target_variable_name}: Not enough data points ({len(X)}).")
        results_summary.append(f"Skipped: Not enough data points ({len(X)}).\n")
        continue

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=42, stratify=y if y.nunique() > 1 else None
    )

    model = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(random_state=42, class_weight='balanced'))
    ])

    model.fit(X_train, y_train)
    y_pred_test = model.predict(X_test)
    y_pred_all = model.predict(X)

    accuracy = accuracy_score(y_test, y_pred_test)
    report = classification_report(y_test, y_pred_test, zero_division=0)

    results_summary.append(f"Test Set Accuracy: {accuracy:.4f}")
    results_summary.append("Test Set Classification Report:")
    results_summary.append(report)
    results_summary.append("\n")

    print(f"Test Accuracy for {target_variable_name}: {accuracy:.4f}")

    predictions_df = X.copy()
    predictions_df[target_variable_name + "_Predicted"] = y_pred_all
    predictions_df["Actual_" + target_variable_name] = y
    predictions_df["Total_Sales_Actual"] = df_agg["Total_Sales"]
    predictions_df["Total_Profit_Actual"] = df_agg["Total_Profit"]

    # Fix SettingWithCopyWarning by copying before assignment
    high_performers = predictions_df[predictions_df[target_variable_name + "_Predicted"] == 1].copy()
    high_performers["Prediction_Type"] = target_variable_name.replace("Is_", "").replace("_Generator", "")
    all_predictions_list.append(high_performers[
        ["Region", "Category", "Sub Category", "Prediction_Type", "Total_Sales_Actual", "Total_Profit_Actual"]
    ])

# Save predicted high performers
if all_predictions_list:
    final_predictions_df = pd.concat(all_predictions_list).drop_duplicates()
    final_predictions_df.sort_values(by=["Region", "Prediction_Type", "Category", "Sub Category"], inplace=True)
    final_predictions_df.to_csv(predicted_high_performers_path, index=False)
    print(f"\nPredicted high-performing combinations saved to {predicted_high_performers_path}")
    results_summary.append(f"Predicted high-performing (Category, Sub Category) combinations per region saved to: {predicted_high_performers_path}\n")
else:
    print("\nNo high performers predicted or models could not be trained.")
    results_summary.append("No high performers predicted or models could not be trained.\n")

# Save summary file
with open(classification_summary_path, "w") as f:
    f.write("Classification Models Summary:\n\n")
    f.write("Objective: Predict (Category, Sub Category) combinations that are likely to be high sales/profit generators within their respective regions.\n")
    f.write("Methodology:\n")
    f.write("1. Aggregated sales and profit for each (Region, Category, Sub Category) combination.\n")
    f.write("2. Defined 'High Sales/Profit Generator' if a combination's total sales/profit was in the top 25% for its region.\n")
    f.write("3. Trained RandomForestClassifier models to predict these binary targets.\n")
    f.write("4. Features used: Region, Category, Sub Category (One-Hot Encoded).\n\n")
    f.write("--- Model Performance & Predictions ---\n")
    for line in results_summary:
        f.write(line + "\n")

print(f"Classification analysis complete. Summary saved to {classification_summary_path}")
```

**Summary of Classification Modeling:**
    Random Forest Classifiers were trained to predict item (Category, Sub Category) combinations that are likely to be high sales or high profit generators within their respective regions.
    - Data was aggregated by Region, Category, and Sub Category.
    - Target variables ('Is_High_Sales_Generator', 'Is_High_Profit_Generator') were created based on the top 25% quantile of sales/profit within each region.
    - Models were trained and evaluated. Predicted high-performing combinations were saved.
    Detailed summary and performance metrics are in `classification_results/classification_summary.txt`.
    Predicted high performers are listed in `classification_results/predicted_high_performers.csv`.

## 6. Data Warehouse Modification (Non-Volatile Feature)

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import os

# Create directory for data warehouse results if it doesn't exist
results_dir = "C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/data_warehouse_modification"
if not os.path.exists(results_dir):
    os.makedirs(results_dir)

dw_summary_path = os.path.join(results_dir, "data_warehouse_modification_summary.txt")
versioned_dataset_path = os.path.join(results_dir, "supermart_sales_versioned.csv")

# Load the original dataset
df_original = pd.read_csv("C:/Users/Shank/Downloads/New folder/DIY_1_RetailSupermarket/Supermart Grocery Sales - Retail Analytics Dataset.csv")

# Convert 'Order Date' to datetime
df_original["Order Date"] = pd.to_datetime(df_original["Order Date"])

print("Original dataset loaded.")

# --- Implement Non-Volatile Feature (Simplified SCD Type 2) --- #

# Define the maximum allowable date for pandas datetime64[ns]
max_pandas_date = pd.to_datetime("2262-04-11")

# Add versioning columns
df_versioned = df_original.copy()
df_versioned["Row_ID"] = range(1, len(df_versioned) + 1) # Unique ID for each original row
df_versioned["Version"] = 1
df_versioned["Effective_Start_Date"] = df_versioned["Order Date"] # Or a fixed historical date
df_versioned["Effective_End_Date"] = max_pandas_date # Represents infinity or currently active
df_versioned["Is_Active"] = True

print("Initial versioning columns added.")

# --- Simulate some updates to demonstrate non-volatility --- #
# For demonstration, we will pick a few records and simulate changes.
# The changes will result in new rows being added, and old rows being marked inactive.

updates_to_apply = [
    {"Order ID": "OD1", "column_to_update": "Discount", "new_value": 0.15, "change_date": "2023-05-15"},
    {"Order ID": "OD2", "column_to_update": "Profit", "new_value": 200.00, "change_date": "2023-06-01"},
    # Add more hypothetical updates here if needed
]

# Keep track of new rows to append
new_rows_list = []

for update in updates_to_apply:
    order_id_to_update = update["Order ID"]
    column_to_update = update["column_to_update"]
    new_value = update["new_value"]
    change_date = pd.to_datetime(update["change_date"])

    current_active_record_index = df_versioned[
        (df_versioned["Order ID"] == order_id_to_update) & (df_versioned["Is_Active"] == True)
    ].index

    if not current_active_record_index.empty:
        idx = current_active_record_index[0]
        
        df_versioned.loc[idx, "Effective_End_Date"] = change_date - timedelta(days=1)
        df_versioned.loc[idx, "Is_Active"] = False
        print(f"Expired record for Order ID {order_id_to_update} (Row_ID: {df_versioned.loc[idx, 'Row_ID']}) on {df_versioned.loc[idx, 'Effective_End_Date']}")

        new_record = df_versioned.loc[idx].copy()
        new_record["Version"] = new_record["Version"] + 1
        new_record[column_to_update] = new_value
        new_record["Effective_Start_Date"] = change_date
        new_record["Effective_End_Date"] = max_pandas_date
        new_record["Is_Active"] = True
        
        new_rows_list.append(new_record)
        print(f"Created new version for Order ID {order_id_to_update} (Row_ID: {new_record['Row_ID']}) with {column_to_update} = {new_value}, effective from {change_date}")
    else:
        print(f"Could not find active record for Order ID {order_id_to_update} to apply update.")

if new_rows_list:
    df_versioned = pd.concat([df_versioned] + [pd.DataFrame([row]) for row in new_rows_list], ignore_index=True)

df_versioned.sort_values(by=["Row_ID", "Version"], inplace=True)

print("Simulated updates applied.")

df_versioned.to_csv(versioned_dataset_path, index=False)
print(f"Versioned dataset saved to {versioned_dataset_path}")

# --- Summary --- #
with open(dw_summary_path, "w") as f:
    f.write("Data Warehouse Modification Summary (Non-Volatile Data):\n\n")
    f.write("Objective: Modify the dataset to incorporate non-volatile features, where data is not updated or deleted but changes are tracked.\n")
    f.write("Methodology (Simplified SCD Type 2):\n")
    f.write("1. Added 'Row_ID' (unique identifier for original rows), 'Version', 'Effective_Start_Date', 'Effective_End_Date', and 'Is_Active' columns.\n")
    f.write(f"2. Initial records are Version 1, Is_Active = True, and Effective_End_Date = {max_pandas_date.strftime('%Y-%m-%d')}.\n")
    f.write("3. Simulated updates for specific 'Order ID's:\n")
    for update in updates_to_apply:
        f.write(f"   - Order ID: {update['Order ID']}, Column: {update['column_to_update']}, New Value: {update['new_value']}, Change Date: {update['change_date']}\n")
    f.write("4. For each update:\n")
    f.write("   a. The existing active record was marked as inactive (Is_Active = False) and its 'Effective_End_Date' was set.\n")
    f.write("   b. A new record was created with the updated value, incremented 'Version', new 'Effective_Start_Date', and Is_Active = True.\n")
    f.write(f"5. The resulting versioned dataset, showing history of changes, is saved to: {versioned_dataset_path}\n\n")
    f.write("Example of changes (first few rows for OD1 and OD2 after versioning if updates applied):\n")
    if "OD1" in df_versioned["Order ID"].values:
        f.write(df_versioned[df_versioned["Order ID"] == "OD1"].to_string() + "\n\n")
    if "OD2" in df_versioned["Order ID"].values:
        f.write(df_versioned[df_versioned["Order ID"] == "OD2"].to_string() + "\n")

print(f"Data warehouse modification process complete. Summary saved to {dw_summary_path}")

```
